```python
# if the loop is completed without break then it goes to else (with break no else part will execute)
n = [1,2,3,4,5]
for i in n:
    if i==3:
        print("found 3, break the loop")
        break
    print(i)
else:
    print("loop is completed without finding a specific condition")
```

```python
# break : to break the loop in between
# continue : to skip particular situation 
# pass : to fill the conditional block temperarly for errorless code

for i in range (10):
    if i==7:
        print("procession stopped")
        #break
        continue
    print(i)
```

```python
# practice
# 1. write a program to perform arithmetic operations according to the given sign
a = int(input("enter value of a :"))
b = int(input("enter value of b :"))
c = input("choose the operator :")
if c == '+':
    print(a+b)
elif c == '-':
    print(a-b)
elif c == '*':
    print(a*b)
elif c == '**':
    print(a**b)
elif c == '/':
    print(a/b)
elif c == '%':
    print(a%b)
elif c == '//':
    print(a//b)
else:
    print("invalid operator")
```
